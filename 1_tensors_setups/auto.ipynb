```python
import torch

x = torch.tensor([1., 2., 3., 4.])
y_true = torch.tensor([2., 4., 6., 8.])    # correct numbers

w = torch.tensor(0.0, requires_grad=True)  # weight parameter

lr = 0.01  # learning rate

epochs = 10 

for epoch in range(epochs):
    # Forward pass: compute predicted y
    y_pred = w * x

    # Compute and print loss
    loss = ((y_pred - y_true) ** 2).mean()  # mean squared error

    # Backward pass: compute gradient of the loss with respect to w
    loss.backward()

    # Update weights using gradient descent
    with torch.no_grad():
        w -= lr * w.grad

    print(f'Epoch {epoch+1}: w = {w.item():.4f}, loss = {loss.item():.4f}')
    
    # Zero the gradients after updating
    w.grad.zero_()
```
